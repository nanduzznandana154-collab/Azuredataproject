```python
presentation_path = "abfss://azuredata@azure00014.dfs.core.windows.net/presentation/"

df_presentation = spark.read.parquet(presentation_path)

display(df_presentation)
```

```python
from pyspark.sql.functions import count, countDistinct, min, max

analysis_summary = df_presentation.agg(
    count("*").alias("total_races"),
    countDistinct("season").alias("total_seasons"),
    countDistinct("circuitId").alias("total_circuits"),
    min("season").alias("first_season"),
    max("season").alias("latest_season")
)

display(analysis_summary)
```

```python
races_by_season = (
    df_presentation
    .groupBy("season")
    .agg(count("*").alias("total_races"))
    .orderBy("season")
)

display(races_by_season)
```

```python
races_by_country = (
    df_presentation
    .groupBy("country")
    .agg(count("*").alias("total_races"))
    .orderBy("total_races", ascending=False)
)

display(races_by_country)
```

```python
races_by_circuit = (
    df_presentation
    .groupBy("circuitName")
    .agg(count("*").alias("total_races"))
    .orderBy("total_races", ascending=False)
)

display(races_by_circuit)
```

```python
races_by_year_country = (
    df_presentation
    .groupBy("season", "country")
    .agg(count("*").alias("total_races"))
    .orderBy("season", "total_races", ascending=False)
)

display(races_by_year_country)
```

```python
circuit_by_season = (
    df_presentation
    .groupBy("season", "circuitName")
    .agg(count("*").alias("total_races"))
    .orderBy("season", "circuitName")
)

display(circuit_by_season)
```
