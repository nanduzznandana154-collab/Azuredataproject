```python
ingested_path = "abfss://azuredata@azure00014.dfs.core.windows.net/ingested/"

df_ingested = spark.read.parquet(ingested_path)

display(df_ingested)
```

```python
from pyspark.sql.functions import col, to_date

df_presentation = (
    df_ingested
    .withColumn("season", col("season").cast("integer"))
    .withColumn("round", col("round").cast("integer"))
    .withColumn("date", to_date(col("date")))
    .withColumn("lat", col("lat").cast("double"))
    .withColumn("long", col("long").cast("double"))
    .dropDuplicates(["season", "round", "raceName"])
)

display(df_presentation)
```

```python
presentation_path = "abfss://azuredata@azure00014.dfs.core.windows.net/presentation/"

(
    df_presentation
    .write
    .mode("overwrite")
    .parquet(presentation_path)
)

print("Presentation data written successfully.")
```
